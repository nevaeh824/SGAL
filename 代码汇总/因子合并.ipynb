```python
# 对治理因子中的 NSAO 进行缺失值填补：
# 1. 如果 CSV 中已有 NSAO 列，则将 NSAO 缺失值填补为 0
# 2. 如果 CSV 中没有 NSAO 列，则新增 NSAO 列，并全部填充为 0

import os
import pandas as pd

# =========================
# 0. 对 治理因子_周频 中所有 CSV 的 NSAO 进行处理
# =========================

governance_input_folder = r'data\因子peer_weekly\治理因子peer_weekly'
governance_folder = r'output\治理因子peer_weekly_filtered'   # 输出到 output，避免覆盖原始数据
os.makedirs(governance_folder, exist_ok=True)

total_rows = 0                  # 所有文件总行数
total_missing_nsao = 0          # 所有文件中，原本已有 NSAO 列时的缺失值总数
total_added_nsao_rows = 0       # 原本没有 NSAO 列、后来新增为 0 的总行数
processed_files = 0             # 成功处理的文件数
failed_files = 0                # 处理失败的文件数
added_nsao_files = 0            # 原本没有 NSAO 列、后来新增 NSAO 列的文件数

print('开始处理 治理因子_周频 文件夹中的 NSAO 列...')

if not os.path.exists(governance_input_folder):
    raise FileNotFoundError(f'治理因子文件夹不存在: {governance_input_folder}')

for file_name in os.listdir(governance_input_folder):
    if not file_name.lower().endswith('.csv'):
        continue

    file_path = os.path.join(governance_input_folder, file_name)
    output_file_path = os.path.join(governance_folder, file_name)

    try:
        df = pd.read_csv(file_path)

        file_total_rows = len(df)
        total_rows += file_total_rows

        # 情况1：如果没有 NSAO 列，则新增 NSAO 列，并全部填充为 0
        if 'NSAO-NSAO_均值' not in df.columns:
            df['NSAO-NSAO_均值'] = 0

            total_added_nsao_rows += file_total_rows
            added_nsao_files += 1

            print(
                f'已处理：{file_name} | '
                f'总行数: {file_total_rows} | '
                f'原本没有 NSAO-NSAO_均值 列，已新增NSAO-NSAO_均值并全部填充为 0'
            )

        # 情况2：如果已有 NSAO 列，则只填补缺失值为 0
        else:
            file_missing_nsao = df['NSAO-NSAO_均值'].isna().sum()
            total_missing_nsao += file_missing_nsao

            df['NSAO-NSAO_均值'] = df['NSAO-NSAO_均值'].fillna(0)

            file_missing_pct = (
                file_missing_nsao / file_total_rows * 100
                if file_total_rows > 0 else 0
            )

            print(
                f'已处理：{file_name} | '
                f'总行数: {file_total_rows} | '
                f'NSAO-NSAO_均值缺失数: {file_missing_nsao} | '
                f'缺失比例: {file_missing_pct:.2f}%'
            )

        # 覆盖保存原 CSV
        df.to_csv(output_file_path, index=False, encoding='utf-8-sig')

        processed_files += 1

    except Exception as e:
        print(f'处理失败：{file_name}，错误原因：{e}')
        failed_files += 1


# 输出整体统计结果
overall_missing_pct = (
    total_missing_nsao / total_rows * 100
    if total_rows > 0 else 0
)

added_nsao_pct = (
    total_added_nsao_rows / total_rows * 100
    if total_rows > 0 else 0
)

print('\nNSAO 列处理完成！')
print(f'成功处理 CSV 文件数: {processed_files}')
print(f'处理失败文件数: {failed_files}')
print(f'原本没有 NSAO 列、已新增的文件数: {added_nsao_files}')
print(f'所有文件总行数: {total_rows}')
print(f'原有 NSAO-NSAO_均值 列中的缺失值数量: {total_missing_nsao}')
print(f'原有 NSAO-NSAO_均值 列中的缺失比例: {overall_missing_pct:.2f}%')
print(f'新增 NSAO-NSAO_均值 并填充为 0 的行数: {total_added_nsao_rows}')
print(f'新增 NSAO-NSAO_均值 行数占总行数比例: {added_nsao_pct:.2f}%')
print('所有 CSV 文件现在都已经包含 NSAO-NSAO_均值 列，且 NSAO-NSAO_均值 中的缺失值已经全部用 0 填补。')
```

```python
import os
import re
import pandas as pd

# =========================
# 1. 路径配置（改成你自己的路径）
# =========================
afco_file = r'data\AF_Co.xlsx'

folder_dict = {
    '财务因子_周频': r'data\因子_weekly\财务因子_周频',
    '情感因子_周频': r'data\因子_weekly\情感因子_周频',
    '市场因子_周频': r'data\因子_weekly\市场因子_周频',
    '治理因子_周频': r'data\因子_weekly\治理因子_周频'
}

# 总汇总输出
summary_output = r'output\股票覆盖对比结果.csv'

# 每个文件夹单独输出
detail_output_dir = r'output\股票覆盖对比明细'
os.makedirs(detail_output_dir, exist_ok=True)


# =========================
# 2. 工具函数
# =========================
def normalize_stock_code(x):
    """
    将股票代码统一规范为6位字符串。
    支持:
    - 1 -> 000001
    - '1' -> 000001
    - '000001' -> 000001
    - '1.0' -> 000001
    """
    if pd.isna(x):
        return None

    s = str(x).strip()

    # 去掉可能的小数形式，如 1.0
    if re.fullmatch(r'\d+\.0+', s):
        s = s.split('.')[0]

    # 提取纯数字
    m = re.search(r'\d+', s)
    if not m:
        return None

    code = m.group()
    return code.zfill(6)


def extract_codes_from_folder(folder_path):
    """
    从文件夹中所有csv文件名提取股票代码，返回set
    例如:
    1.csv -> 000001
    000001.csv -> 000001
    """
    code_set = set()

    if not os.path.exists(folder_path):
        print(f'文件夹不存在: {folder_path}')
        return code_set

    for root, _, files in os.walk(folder_path):
        for file_name in files:
            if not file_name.lower().endswith('.csv'):
                continue

            # 去掉扩展名
            base_name = os.path.splitext(file_name)[0].strip()

            code = normalize_stock_code(base_name)
            if code is not None:
                code_set.add(code)
            else:
                print(f'无法从文件名提取股票代码，已跳过: {file_name}')

    return code_set


def find_code_column(df):
    """
    自动寻找AF_Co.xlsx中的股票代码列
    优先匹配常见列名
    """
    candidate_cols = ['证券代码', '股票代码', '代码', 'stock_code', 'code', 'Stkcd']

    for col in candidate_cols:
        if col in df.columns:
            return col

    # 如果没有找到常见列名，尝试找包含“代码”的列
    for col in df.columns:
        if '代码' in str(col):
            return col

    raise ValueError(
        f'未能在 AF_Co.xlsx 中找到股票代码列，请检查列名。当前列名为: {list(df.columns)}'
    )


def pad_list_to_same_length(dict_of_lists):
    """
    将不同长度的列表补齐，方便写入同一个DataFrame
    """
    max_len = max(len(v) for v in dict_of_lists.values()) if dict_of_lists else 0
    padded = {}
    for k, v in dict_of_lists.items():
        padded[k] = v + [None] * (max_len - len(v))
    return padded


# =========================
# 3. 读取 AF_Co.xlsx 中的完整股票代码
# =========================
afco = pd.read_excel(afco_file)
code_col = find_code_column(afco)

afco_codes = (
    afco[code_col]
    .apply(normalize_stock_code)
    .dropna()
    .drop_duplicates()
)

afco_code_set = set(afco_codes)

print(f'AF_Co.xlsx 总股票数（去重后）: {len(afco_code_set)}')


# =========================
# 4. 逐个文件夹对比
# =========================
summary_dict = {}
stat_rows = []

for folder_name, folder_path in folder_dict.items():
    folder_codes = extract_codes_from_folder(folder_path)

    # AF_Co里有，但该文件夹里没有
    missing_in_folder = sorted(afco_code_set - folder_codes)

    # 该文件夹里有，但AF_Co里没有
    extra_in_folder = sorted(folder_codes - afco_code_set)

    # 交集
    common_codes = sorted(afco_code_set & folder_codes)

    print(f'\n[{folder_name}]')
    print(f'文件夹股票数: {len(folder_codes)}')
    print(f'与AF_Co重合数: {len(common_codes)}')
    print(f'AF_Co中有但文件夹缺少: {len(missing_in_folder)}')
    print(f'文件夹中有但AF_Co没有: {len(extra_in_folder)}')

    # 保存统计信息
    stat_rows.append({
        '文件夹名称': folder_name,
        'AF_Co股票数': len(afco_code_set),
        '文件夹股票数': len(folder_codes),
        '重合股票数': len(common_codes),
        'AF_Co中有但文件夹缺少': len(missing_in_folder),
        '文件夹中有但AF_Co没有': len(extra_in_folder)
    })

    # 保存每个文件夹的详细结果
    detail_df = pd.DataFrame({
        'AF_Co中有但该文件夹缺少': pd.Series(missing_in_folder),
        '该文件夹中有但AF_Co没有': pd.Series(extra_in_folder),
        'AF_Co与该文件夹共有': pd.Series(common_codes)
    })

    detail_file = os.path.join(detail_output_dir, f'{folder_name}_对比明细.csv')
    detail_df.to_csv(detail_file, index=False, encoding='utf-8-sig')

    # 用于总汇总表
    summary_dict[f'{folder_name}_AF_Co中有但缺少'] = missing_in_folder
    summary_dict[f'{folder_name}_文件夹中有但AF_Co没有'] = extra_in_folder


# =========================
# 5. 输出总汇总结果
# =========================
summary_padded = pad_list_to_same_length(summary_dict)
summary_df = pd.DataFrame(summary_padded)
summary_df.to_csv(summary_output, index=False, encoding='utf-8-sig')

# 统计表也单独保存
stat_df = pd.DataFrame(stat_rows)
stat_df.to_csv(os.path.join(detail_output_dir, '各文件夹覆盖统计.csv'),
               index=False, encoding='utf-8-sig')

print('\n处理完成！')
print(f'总汇总文件已保存: {summary_output}')
print(f'各文件夹明细已保存到: {detail_output_dir}')
```

```python
import os
import re
import pandas as pd

# =========================
# 1. 路径设置（改成你自己的）
# =========================
afco_file = r'data\AF_Co.xlsx'

finance_folder = r'data\因子_weekly\财务因子_周频'
sentiment_folder = r'data\因子_weekly\情感因子_周频'
market_folder = r'data\因子_weekly\市场因子_周频'
governance_folder = r'data\因子_weekly\治理因子_周频'

output_file = r'output\股票代码按行对齐表.csv'


# =========================
# 2. 股票代码标准化函数
# =========================
def normalize_stock_code(x):
    """
    将股票代码统一转为6位字符串
    例如:
    1 -> 000001
    '1' -> 000001
    '000001' -> 000001
    '1.0' -> 000001
    """
    if pd.isna(x):
        return None

    s = str(x).strip()

    # 处理 excel 里可能出现的 1.0
    if re.fullmatch(r'\d+\.0+', s):
        s = s.split('.')[0]

    # 提取数字
    m = re.search(r'\d+', s)
    if not m:
        return None

    return m.group().zfill(6)


# =========================
# 3. 自动识别 AF_CO 中的股票代码列
# =========================
def find_code_column(df):
    candidate_cols = ['证券代码', '股票代码', '代码', 'stock_code', 'code', 'Stkcd']
    for col in candidate_cols:
        if col in df.columns:
            return col

    for col in df.columns:
        if '代码' in str(col):
            return col

    raise ValueError(f'找不到股票代码列，请检查 AF_Co.xlsx 的列名：{list(df.columns)}')


# =========================
# 4. 读取 AF_CO 股票代码
# =========================
afco_df = pd.read_excel(afco_file)
code_col = find_code_column(afco_df)

afco_set = set(
    afco_df[code_col]
    .apply(normalize_stock_code)
    .dropna()
    .tolist()
)


# =========================
# 5. 从文件夹文件名中提取股票代码
# =========================
def extract_codes_from_folder(folder_path):
    code_set = set()

    if not os.path.exists(folder_path):
        print(f'文件夹不存在：{folder_path}')
        return code_set

    for root, _, files in os.walk(folder_path):
        for file_name in files:
            if not file_name.lower().endswith('.csv'):
                continue

            base_name = os.path.splitext(file_name)[0].strip()
            code = normalize_stock_code(base_name)

            if code is not None:
                code_set.add(code)
            else:
                print(f'无法识别的文件名，已跳过：{file_name}')

    return code_set


finance_set = extract_codes_from_folder(finance_folder)
sentiment_set = extract_codes_from_folder(sentiment_folder)
market_set = extract_codes_from_folder(market_folder)
governance_set = extract_codes_from_folder(governance_folder)


# =========================
# 6. 构造“按股票代码对齐”的结果表
#    以所有股票代码的并集为基准
# =========================
all_codes = sorted(
    afco_set | finance_set | sentiment_set | market_set | governance_set
)

result_rows = []
for code in all_codes:
    result_rows.append({
        '股票代码': code,
        'AF_CO': code if code in afco_set else '',
        '财务因子': code if code in finance_set else '',
        '情感因子': code if code in sentiment_set else '',
        '市场因子': code if code in market_set else '',
        '治理因子': code if code in governance_set else '',
    })

result_df = pd.DataFrame(result_rows)


# =========================
# 7. 输出
# =========================
result_df.to_csv(output_file, index=False, encoding='utf-8-sig')

print('输出完成：', output_file)
print('AF_CO 数量:', len(afco_set))
print('财务因子 数量:', len(finance_set))
print('情感因子 数量:', len(sentiment_set))
print('市场因子 数量:', len(market_set))
print('治理因子 数量:', len(governance_set))
print('并集股票数量:', len(all_codes))
```

```python
# ============================================================
# 检查同一股票同一年内 week 是否连续
# week 格式如：201501，表示 2015 年第 1 周
# 只检查一年内部，不检查跨年是否连续
# ============================================================

import pandas as pd
FSFP_PATH = r"output\fsfp_label.csv"
stock_col = "code"
# 建议读取 code 时保留字符串，否则 000001 会变成 1
df = pd.read_csv(FSFP_PATH, dtype={stock_col: str})
# 如果你希望股票代码显示成 6 位
df[stock_col] = df[stock_col].astype(str).str.zfill(6)

# 1. 处理 week 列
df["week"] = pd.to_numeric(df["week"], errors="coerce")

# 检查 week 是否存在无法转成数字的情况
invalid_week_format_rows = df[df["week"].isna()]

print(f"week 无法转换为数字的记录数: {len(invalid_week_format_rows)}")

if len(invalid_week_format_rows) > 0:
    print("警告：存在 week 格式无法转换为数字的记录，下面展示部分样本：")
    display(invalid_week_format_rows[[stock_col, "week"]].head(20))

# 只对 week 有效的记录继续检查
df_valid = df[df["week"].notna()].copy()
df_valid["week"] = df_valid["week"].astype(int)

# 2. 拆分 year 和 week_num
df_valid["_check_year"] = df_valid["week"] // 100
df_valid["_check_week_num"] = df_valid["week"] % 100

# 3. 检查 week_num 是否在 1-53 范围内
invalid_week_rows = df_valid[
    (df_valid["_check_week_num"] < 1) | 
    (df_valid["_check_week_num"] > 53)
]

print(f"week_num 不在 1-53 范围内的记录数: {len(invalid_week_rows)}")

if len(invalid_week_rows) > 0:
    print("警告：存在 week_num 异常的记录，下面展示部分样本：")
    display(
        invalid_week_rows[
            [stock_col, "week", "_check_year", "_check_week_num"]
        ].head(20)
    )

# 只保留 week_num 合理的记录用于连续性检查
df_check = df_valid[
    (df_valid["_check_week_num"] >= 1) & 
    (df_valid["_check_week_num"] <= 53)
].copy()

# 4. 去重，避免同一股票同一年同一周重复记录影响判断
week_check_df = (
    df_check[[stock_col, "week", "_check_year", "_check_week_num"]]
    .drop_duplicates(subset=[stock_col, "_check_year", "_check_week_num"])
    .sort_values([stock_col, "_check_year", "_check_week_num"])
    .copy()
)

# 5. 计算同一股票、同一年内部的上一周和下一周
week_check_df["_prev_week"] = (
    week_check_df
    .groupby([stock_col, "_check_year"])["week"]
    .shift(1)
)

week_check_df["_next_week"] = (
    week_check_df
    .groupby([stock_col, "_check_year"])["week"]
    .shift(-1)
)

week_check_df["_prev_week_num"] = (
    week_check_df
    .groupby([stock_col, "_check_year"])["_check_week_num"]
    .shift(1)
)

week_check_df["_next_week_num"] = (
    week_check_df
    .groupby([stock_col, "_check_year"])["_check_week_num"]
    .shift(-1)
)

# 当前周 - 上一周
week_check_df["_week_diff_from_prev"] = (
    week_check_df["_check_week_num"] - week_check_df["_prev_week_num"]
)

# 下一周 - 当前周
week_check_df["_week_diff_to_next"] = (
    week_check_df["_next_week_num"] - week_check_df["_check_week_num"]
)

# 6. 找出和上一条记录不连续的行
gap_rows = week_check_df[
    week_check_df["_week_diff_from_prev"].notna()
    & (week_check_df["_week_diff_from_prev"] != 1)
].copy()

# 7. 生成缺失周说明，不使用 apply，避免你的报错
missing_weeks_list = []
missing_week_count_list = []

for _, row in gap_rows.iterrows():
    year = int(row["_check_year"])
    prev_week_num = int(row["_prev_week_num"])
    curr_week_num = int(row["_check_week_num"])

    missing_weeks = list(range(prev_week_num + 1, curr_week_num))

    missing_week_count_list.append(len(missing_weeks))

    if len(missing_weeks) == 0:
        missing_weeks_list.append("")
    else:
        missing_weeks_list.append(
            ", ".join([f"{year}{w:02d}" for w in missing_weeks])
        )

gap_rows["_missing_week_count"] = missing_week_count_list
gap_rows["_missing_weeks"] = missing_weeks_list

# 8. 输出检查结果
print(f"同一股票同一年内 week 不连续的断点记录数: {len(gap_rows)}")

if len(gap_rows) > 0:
    print(f"估计缺失的 股票-周 组合总数: {gap_rows['_missing_week_count'].sum()}")
    print("警告：存在同一股票同一年内 week 不连续的情况，下面展示部分样本：")

    display(
        gap_rows[
            [
                stock_col,
                "_check_year",
                "_prev_week",
                "week",
                "_next_week",
                "_prev_week_num",
                "_check_week_num",
                "_next_week_num",
                "_week_diff_from_prev",
                "_week_diff_to_next",
                "_missing_week_count",
                "_missing_weeks"
            ]
        ].head(30)
    )
else:
    print("检查通过：同一股票同一年内 week 是连续的。")
```

```python
import os
import re
import glob
import pandas as pd
from functools import reduce

# =========================
# 1. 路径配置
# =========================
FSFP_PATH = r"output\FSFP_next_cts_0509.csv"
FACTOR_ROOT = r"."
FACTOR_FOLDERS = [
    r"data\因子_weekly\财务因子_周频",
    r"data\因子_weekly\治理因子_周频",
    r"data\因子_weekly\市场因子_周频",
    r"data\因子_weekly\情感因子_周频",
    r"data\因子peer_weekly\财务因子peer_weekly",
    r"data\因子peer_weekly\市场因子peer_weekly",
    r"data\因子peer_weekly\治理因子peer_weekly",
    r"data\因子peer_weekly\情感因子peer_weekly"
]

OUTPUT_PATH = r"output\final_merged_inner_cts_0509.csv"
ID_COL = "code"
TIME_COL = "week"
TARGET_COL = "FSFP_next"


# =========================
# 2. 基础工具函数
# =========================
def normalize_code(x):
    """
    统一股票代码为6位字符串
    兼容:
    1 / 1.0 / '1' / '000001' / '000001.SZ'
    """
    if pd.isna(x):
        return None

    s = str(x).strip()

    # 提取数字部分
    nums = re.findall(r"\d+", s)
    if not nums:
        return None

    code = nums[0].zfill(6)
    return code


def normalize_week(x):
    """
    统一 week 为整数，如 201501
    """
    if pd.isna(x):
        return None
    try:
        s = str(x).strip()
        # 去掉可能的 .0
        s = s.replace(".0", "")
        return int(s)
    except:
        return None


def get_file_code(file_path):
    """
    从文件名提取股票代码
    例如:
    1.csv -> 000001
    000001.csv -> 000001
    """
    base = os.path.basename(file_path)
    name = os.path.splitext(base)[0]
    return normalize_code(name)


def find_col(df, candidates):
    """
    在 df.columns 中查找候选列名（忽略大小写）
    """
    lower_map = {str(c).strip().lower(): c for c in df.columns}
    for c in candidates:
        if c.lower() in lower_map:
            return lower_map[c.lower()]
    return None


def safe_read_csv(file_path):
    """
    尽量稳妥地读取 csv
    """
    encodings = ["utf-8-sig", "utf-8", "gbk", "gb18030"]
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc, low_memory=False)
        except:
            continue
    # 最后直接再试一次默认编码，失败就抛错
    return pd.read_csv(file_path, low_memory=False)


# =========================
# 3. 读取 FSFP
# =========================
def load_fsfp(fsfp_path):
    print(f"正在读取 FSFP 文件: {fsfp_path}")
    df = pd.read_csv(fsfp_path, low_memory=False)

    code_col = find_col(df, ["code", "Stkcd", "stock_code"])
    week_col = find_col(df, ["week", "Week", "trdweek"])
    fsfp_col = find_col(df, ["FSFP_next"])

    if code_col is None:
        raise ValueError("FSFP 文件中未找到股票代码列（code / Stkcd / stock_code）")
    if week_col is None:
        raise ValueError("FSFP 文件中未找到时间列（week / Week / trdweek）")
    if fsfp_col is None:
        raise ValueError("FSFP 文件中未找到 FSFP_next 列")

    df = df[[code_col, week_col, fsfp_col]].copy()
    df.columns = [ID_COL, TIME_COL, TARGET_COL]

    df[ID_COL] = df[ID_COL].apply(normalize_code)
    df[TIME_COL] = df[TIME_COL].apply(normalize_week)
    df[TARGET_COL] = pd.to_numeric(df[TARGET_COL], errors="coerce")

    df = df.dropna(subset=[ID_COL, TIME_COL, TARGET_COL])
    df[TIME_COL] = df[TIME_COL].astype(int)
    df[TARGET_COL] = df[TARGET_COL].astype(int)

    # 同一 (code, week) 去重
    df = df.drop_duplicates(subset=[ID_COL, TIME_COL], keep="first").reset_index(drop=True)

    print(f"FSFP 读取完成，形状: {df.shape}")
    print(f"FSFP 股票数: {df[ID_COL].nunique()}")
    return df


# =========================
# 4. 扫描每个文件夹拥有的股票代码
# =========================
def scan_folder_codes(folder_path):
    """
    通过文件名扫描该文件夹下有哪些股票代码
    """
    files = glob.glob(os.path.join(folder_path, "**", "*.csv"), recursive=True)
    codes = set()

    for f in files:
        code = get_file_code(f)
        if code is not None:
            codes.add(code)

    return codes


# =========================
# 5. 读取单个文件夹因子
# =========================
def load_factor_folder(folder_path, common_codes, prefix):
    """
    读取一个文件夹下的所有因子文件，只保留 common_codes
    输出统一格式:
    code, week, 各因子列
    """
    print(f"\n正在读取文件夹: {folder_path}")
    files = glob.glob(os.path.join(folder_path, "**", "*.csv"), recursive=True)

    all_dfs = []

    for f in files:
        file_code = get_file_code(f)
        if file_code is None or file_code not in common_codes:
            continue

        try:
            tmp = safe_read_csv(f)

            code_col = find_col(tmp, ["code", "Stkcd", "stock_code"])
            week_col = find_col(tmp, ["week", "Week", "trdweek"])

            if week_col is None:
                print(f"  [跳过] 未找到 week 列: {os.path.basename(f)}")
                continue

            # 如果文件里有 code/Stkcd，则优先用文件内的；否则用文件名
            if code_col is not None:
                tmp[ID_COL] = tmp[code_col].apply(normalize_code)
            else:
                tmp[ID_COL] = file_code

            tmp[TIME_COL] = tmp[week_col].apply(normalize_week)

            # 删掉无效行
            tmp = tmp.dropna(subset=[ID_COL, TIME_COL]).copy()
            tmp[TIME_COL] = tmp[TIME_COL].astype(int)

            # 删除原始ID/时间列，避免重复
            drop_cols = set()
            if code_col is not None:
                drop_cols.add(code_col)
            if week_col is not None:
                drop_cols.add(week_col)
            if "fsfp_original" in tmp.columns:
                drop_cols.add("fsfp_original")
            if "FSFP" in tmp.columns:
                drop_cols.add("FSFP")

            tmp = tmp.drop(columns=[c for c in drop_cols if c in tmp.columns], errors="ignore")

            # 把因子列加前缀，避免重名冲突
            rename_dict = {}
            for c in tmp.columns:
                if c not in [ID_COL, TIME_COL]:
                    rename_dict[c] = f"{prefix}__{c}"
            tmp = tmp.rename(columns=rename_dict)

            # 同一文件内去重
            tmp = tmp.drop_duplicates(subset=[ID_COL, TIME_COL], keep="first")

            all_dfs.append(tmp)

        except Exception as e:
            print(f"  [跳过] 读取失败: {os.path.basename(f)} | 原因: {e}")
            continue

    if not all_dfs:
        print(f"  [警告] 文件夹无有效数据: {folder_path}")
        return pd.DataFrame(columns=[ID_COL, TIME_COL])

    folder_df = pd.concat(all_dfs, ignore_index=True)

    # 同一文件夹内，如果同一 (code, week) 出现多条，按 first 合并
    folder_df = folder_df.groupby([ID_COL, TIME_COL], as_index=False).first()

    print(f"文件夹读取完成: {os.path.basename(folder_path)} | 形状: {folder_df.shape}")
    print(f"股票数: {folder_df[ID_COL].nunique()} | 时间点数: {folder_df[TIME_COL].nunique()}")
    return folder_df


# =========================
# 6. 主程序
# =========================
def main():
    # 1) 读取 FSFP
    df_fsfp = load_fsfp(FSFP_PATH)
    fsfp_codes = set(df_fsfp[ID_COL].unique())

    # 2) 扫描每个文件夹的股票代码
    folder_code_sets = {}
    for folder in FACTOR_FOLDERS:
        folder_path = os.path.join(FACTOR_ROOT, folder)
        if not os.path.exists(folder_path):
            raise FileNotFoundError(f"文件夹不存在: {folder_path}")

        codes = scan_folder_codes(folder_path)
        folder_code_sets[folder] = codes
        print(f"{folder} 文件数对应股票数: {len(codes)}")

    # 3) 求共同股票交集：FSFP 和 8 个文件夹都必须存在
    common_codes = fsfp_codes.copy()
    for folder in FACTOR_FOLDERS:
        common_codes &= folder_code_sets[folder]

    common_codes = sorted(common_codes)

    print("\n============================")
    print(f"共同股票数量: {len(common_codes)}")
    print("============================\n")

    if len(common_codes) == 0:
        raise ValueError("FSFP 与 8 个文件夹没有共同股票，无法合并。")

    # 4) FSFP 先裁剪到共同股票
    df_fsfp = df_fsfp[df_fsfp[ID_COL].isin(common_codes)].copy()

    # 5) 读取每个文件夹的因子
    folder_prefix_map = {
        r"data\因子_weekly\财务因子_周频": "fin",
        r"data\因子_weekly\治理因子_周频": "gov",
        r"data\因子_weekly\市场因子_周频": "mkt",
        r"data\因子_weekly\情感因子_周频": "sent",
        r"data\因子peer_weekly\财务因子peer_weekly": "peer_fin",
        r"data\因子peer_weekly\市场因子peer_weekly": "peer_mkt",
        r"data\因子peer_weekly\治理因子peer_weekly": "peer_gov",
        r"data\因子peer_weekly\情感因子peer_weekly": "peer_sent"
    }

    factor_dfs = []
    for folder in FACTOR_FOLDERS:
        folder_path = os.path.join(FACTOR_ROOT, folder)
        prefix = folder_prefix_map.get(folder, folder)
        df_folder = load_factor_folder(folder_path, common_codes, prefix)
        factor_dfs.append(df_folder)

    # 6) 逐步 inner merge
    print("\n开始执行 inner merge ...")
    merged_df = df_fsfp.copy()

    for i, df_factor in enumerate(factor_dfs, start=1):
        print(f"正在合并第 {i}/{len(factor_dfs)} 个文件夹 ...")
        before_shape = merged_df.shape
        merged_df = pd.merge(
            merged_df,
            df_factor,
            on=[ID_COL, TIME_COL],
            how="inner"
        )
        after_shape = merged_df.shape
        print(f"  合并前: {before_shape} -> 合并后: {after_shape}")

    # 7) 排序
    merged_df = merged_df.sort_values([ID_COL, TIME_COL]).reset_index(drop=True)

    # 8) 输出
    print("\n最终结果：")
    print(f"形状: {merged_df.shape}")
    print(f"股票数: {merged_df[ID_COL].nunique()}")
    print(f"时间点数: {merged_df[TIME_COL].nunique()}")

    merged_df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8-sig")
    print(f"\n已保存到: {OUTPUT_PATH}")


if __name__ == "__main__":
    main()
```

```python
# 合并 FSFP_peer

import os
import re
import pandas as pd

# =========================
# 1. 路径配置
# =========================
BASE_MERGED_PATH = r"output\final_merged_inner_cts_0509.csv"
PEER_FSFP_PATH = r"data\因子peer_weekly\FSFP__词频_peer\peer_fsfp_weekly_0509.csv"
OUTPUT_PATH = r"output\final_merged_0509_cts.csv"

ID_COL = "code"
TIME_COL = "week"

# 新增单文件因子列名
NEW_SINGLE_FACTOR_TARGET = "Peer_FSFP"


# =========================
# 2. 基础工具函数
# =========================
def normalize_code(x):
    """
    统一股票代码为6位字符串
    兼容:
    1 / 1.0 / '1' / '000001' / '000001.SZ'
    """
    if pd.isna(x):
        return None

    s = str(x).strip()
    nums = re.findall(r"\d+", s)
    if not nums:
        return None

    code = nums[0].zfill(6)
    return code


def normalize_week(x):
    """
    统一 week 为整数，如 201501
    """
    if pd.isna(x):
        return None

    try:
        s = str(x).strip()
        s = s.replace(".0", "")
        return int(s)
    except Exception:
        return None


def find_col(df, candidates):
    """
    在 df.columns 中查找候选列名，忽略大小写
    """
    lower_map = {str(c).strip().lower(): c for c in df.columns}

    for c in candidates:
        if c.lower() in lower_map:
            return lower_map[c.lower()]

    return None


def safe_read_csv(file_path):
    """
    尽量稳妥地读取 csv
    """
    encodings = ["utf-8-sig", "utf-8", "gbk", "gb18030"]

    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc, low_memory=False)
        except Exception:
            continue

    return pd.read_csv(file_path, low_memory=False)


# =========================
# 3. 读取已有拼接结果
# =========================
def load_base_merged(base_path):
    print(f"正在读取历史拼接结果: {base_path}")
    df = safe_read_csv(base_path)

    code_col = find_col(df, ["code", "Stkcd", "stock_code"])
    week_col = find_col(df, ["week", "Week", "trdweek"])

    if code_col is None:
        raise ValueError("历史拼接文件中未找到股票代码列（code / Stkcd / stock_code）")

    if week_col is None:
        raise ValueError("历史拼接文件中未找到时间列（week / Week / trdweek）")

    rename_map = {}

    if code_col != ID_COL:
        rename_map[code_col] = ID_COL

    if week_col != TIME_COL:
        rename_map[week_col] = TIME_COL

    if rename_map:
        df = df.rename(columns=rename_map)

    df[ID_COL] = df[ID_COL].apply(normalize_code)
    df[TIME_COL] = df[TIME_COL].apply(normalize_week)

    df = df.dropna(subset=[ID_COL, TIME_COL]).copy()
    df[TIME_COL] = df[TIME_COL].astype(int)

    df = df.drop_duplicates(
        subset=[ID_COL, TIME_COL],
        keep="first"
    ).reset_index(drop=True)

    print(f"历史拼接结果读取完成，形状: {df.shape}")
    print(f"股票数: {df[ID_COL].nunique()} | 时间点数: {df[TIME_COL].nunique()}")

    return df


# =========================
# 4. 读取新增 peer_fsfp 单文件
# =========================
def load_peer_fsfp(peer_fsfp_path):
    print(f"\n正在读取新增单文件因子: {peer_fsfp_path}")
    df = safe_read_csv(peer_fsfp_path)

    code_col = find_col(df, ["code", "Stkcd", "stock_code"])
    week_col = find_col(df, ["week", "Week", "trdweek"])
    factor_col = find_col(df, ["Peer_FSFP", "peer_fsfp", "FSFP"])

    if code_col is None:
        raise ValueError("peer_fsfp 文件中未找到股票代码列（code / Stkcd / stock_code）")

    if week_col is None:
        raise ValueError("peer_fsfp 文件中未找到时间列（week / Week / trdweek）")

    if factor_col is None:
        raise ValueError("peer_fsfp 文件中未找到目标因子列（Peer_FSFP / peer_fsfp / FSFP）")

    df = df[[code_col, week_col, factor_col]].copy()
    df.columns = [ID_COL, TIME_COL, NEW_SINGLE_FACTOR_TARGET]

    df[ID_COL] = df[ID_COL].apply(normalize_code)
    df[TIME_COL] = df[TIME_COL].apply(normalize_week)

    df = df.dropna(subset=[ID_COL, TIME_COL, NEW_SINGLE_FACTOR_TARGET]).copy()
    df[TIME_COL] = df[TIME_COL].astype(int)

    df = df.drop_duplicates(
        subset=[ID_COL, TIME_COL],
        keep="first"
    ).reset_index(drop=True)

    print(f"新增单文件因子读取完成，形状: {df.shape}")
    print(f"股票数: {df[ID_COL].nunique()} | 时间点数: {df[TIME_COL].nunique()}")

    return df


# =========================
# 5. 主程序
# =========================
def main():
    # 0) 路径检查
    if not os.path.exists(BASE_MERGED_PATH):
        raise FileNotFoundError(f"历史拼接文件不存在: {BASE_MERGED_PATH}")

    if not os.path.exists(PEER_FSFP_PATH):
        raise FileNotFoundError(f"新增单文件不存在: {PEER_FSFP_PATH}")

    # 1) 读取历史拼接结果
    df_base = load_base_merged(BASE_MERGED_PATH)

    # 2) 读取新增 Peer_FSFP 单文件
    df_peer_fsfp = load_peer_fsfp(PEER_FSFP_PATH)

    # 3) 求共同股票交集
    base_codes = set(df_base[ID_COL].unique())
    peer_fsfp_codes = set(df_peer_fsfp[ID_COL].unique())

    common_codes = sorted(base_codes & peer_fsfp_codes)

    print("\n============================")
    print(f"共同股票数量: {len(common_codes)}")
    print("============================\n")

    if len(common_codes) == 0:
        raise ValueError("历史拼接结果与 peer_fsfp 单文件没有共同股票，无法合并。")

    # 4) 裁剪到共同股票
    df_base = df_base[df_base[ID_COL].isin(common_codes)].copy()
    df_peer_fsfp = df_peer_fsfp[df_peer_fsfp[ID_COL].isin(common_codes)].copy()

    # 5) inner merge
    print("开始执行 inner merge ...")
    before_shape = df_base.shape

    merged_df = pd.merge(
        df_base,
        df_peer_fsfp,
        on=[ID_COL, TIME_COL],
        how="inner"
    )

    after_shape = merged_df.shape
    print(f"合并前: {before_shape} -> 合并后: {after_shape}")

    # 6) 排序
    merged_df = merged_df.sort_values([ID_COL, TIME_COL]).reset_index(drop=True)

    # 7) 输出结果
    print("\n最终结果：")
    print(f"形状: {merged_df.shape}")
    print(f"股票数: {merged_df[ID_COL].nunique()}")
    print(f"时间点数: {merged_df[TIME_COL].nunique()}")

    merged_df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8-sig")

    print(f"\n已保存到: {OUTPUT_PATH}")


if __name__ == "__main__":
    main()
```
